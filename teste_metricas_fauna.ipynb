# 📊 Teste e Validação de Métricas - Dashboard da Fauna Brasileira
Este notebook tem como objetivo carregar os dados brutos do ICMBio (DwC-A), realizar o cruzamento relacional e validar matematicamente todas as métricas de negócio e KPIs exibidos no Power BI.

```python
import pandas as pd
import numpy as np

print("Bibliotecas carregadas com sucesso!")
```

```python
# Carregamento dos ficheiros de dados originais
df_taxon = pd.read_csv("taxon.txt", sep="\t", low_memory=False)
df_dist = pd.read_csv("distribution.txt", sep="\t", low_memory=False)

# Cruzamento das bases através da chave primária 'id' (taxon_id) e limpeza de duplicados
df_merged = pd.merge(
    df_dist[['id', 'countryCode', 'threatStatus']],
    df_taxon[['id', 'scientificName', 'kingdom', 'phylum', 'class', 'order', 'family', 'taxonRank']],
    on='id',
    how='inner'
).drop_duplicates(subset=['id'])

print(f"Total de registos consolidados na base única: {df_merged.shape[0]}")
```

```python
# Validação da Métrica: Total Geral de Espécies 
total_geral_especies = df_merged['id'].nunique()

print(f"==================================================")
print(f" TOTAL GERAL DE ESPÉCIES: {total_geral_especies}")
print(f"==================================================")
```

```python
# Validação das Métricas de Status de Ameaça (CR, EN, VU)

total_cr = df_merged[df_merged['threatStatus'] == 'CR']['id'].nunique()
total_en = df_merged[df_merged['threatStatus'] == 'EN']['id'].nunique()
total_vu = df_merged[df_merged['threatStatus'] == 'VU']['id'].nunique()

print("==================================================")
print(" MÉTRICAS DE STATUS DE AMEAÇA")
print("==================================================")
print(f"• Espécies Críticas (CR): {total_cr}")
print(f"• Espécies Em Perigo (EN): {total_en}")
print(f"• Espécies Vulneráveis (VU): {total_vu}")

# Validação percentual
print("\n--- Percentuais sobre o Total ---")
print(f"• CR: {(total_cr / total_geral_especies) * 100:.2f}%")
print(f"• EN: {(total_en / total_geral_especies) * 100:.2f}%")
print(f"• VU: {(total_vu / total_geral_especies) * 100:.2f}%")
```

```python
# Validação da Distribuição por Filo e Classe 

print("==================================================")
print(" TOP 5 FILOS MAIS REPRESENTADOS")
print("==================================================")
top_filos = df_merged['phylum'].value_counts().head(5)
print(top_filos)

print("\n==================================================")
print(" TOP 5 CLASSES MAIS REPRESENTADAS")
print("==================================================")
top_classes = df_merged['class'].value_counts().head(5)
print(top_classes)
```
